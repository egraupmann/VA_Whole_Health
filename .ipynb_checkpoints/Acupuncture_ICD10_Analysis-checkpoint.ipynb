## Establish the database connections with SQL Server
The recommended workflow is to make your data tables in Microsoft SQL Server and then simply access the joined/filtered outputs here. Trying to query the full datasets in Python is likely to create memory timeouts, depending on how much data you're trying to pull back. Creating and saving the desired manipulated data in SQL Server also enables others to more easily access your data products.

```python
%%time
# establish the connection parameters
odbc_str = ("DRIVER={ODBC Driver 18 for SQL Server};"
            "SERVER=VHACDWA01.vha.med.va.gov,1433;"
            "DATABASE=OPCCCT_CIH;"
            "Trusted_Connection=yes;"
            "Encrypt=yes;"
            "TrustServerCertificate=no;")

# connect
params = urllib.parse.quote_plus(odbc_str)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")
```

# Get ICD 10 counts based on different encounter slices

```python
%%time
# get the time filters
today = pd.Timestamp.today().normalize()
two_years_ago = today - pd.DateOffset(years=2)

# get encounter and unique patient counts by ICD10
def get_ICD10_counts(df, summ, icd_cnt_fn, pat_cnt_fn):
    
    # counts
    if summ==1: # for the counts not filtered by time
        df[icd_cnt_fn] = df.groupby(['ICD10Code'])['TimesDiagnosedTotal'].transform('sum')
        df[pat_cnt_fn] = df.groupby(['ICD10Code'])['PatientICN'].transform('count')

    else:
        df = df[(df['DiagnosisDateTime'] >= two_years_ago) & (df['DiagnosisDateTime'] <= today)]
        df[icd_cnt_fn] = df.groupby(['ICD10Code'])['PatientICN'].transform('count')
        df[pat_cnt_fn] = df.groupby(['ICD10Code'])['PatientICN'].transform('nunique')


    df = df.filter(['ICD10Code', icd_cnt_fn, pat_cnt_fn]).drop_duplicates()
    
    # remove ICD10s w/ virtually not diagnoses
    df = df.loc[df[pat_cnt_fn]>100]
    df = df.sort_values(by=[icd_cnt_fn], ascending=False)

    return(df)


# For acupuncture specific encounters since FY21
icd_acu = pd.read_sql("SELECT * FROM [OPCCCT_CIH].[dflt].[acupuncture_diagnoses_summary_acu_only]", engine)
icd_acu = get_ICD10_counts(icd_acu, 1, 'icd10_count_acu_all', 'icd10_patient_count_acu_all')
print('done with acupuncture diagnosis summary')

# For acupuncture specific encounters in the last 24 months
icd_acu_24 = pd.read_sql("SELECT * FROM [OPCCCT_CIH].[dflt].[acupuncture_all_diagnoses_acu_only]", engine)
icd_acu_24 = get_ICD10_counts(icd_acu_24, 0, 'icd10_count_acu_24', 'icd10_patient_count_acu_24')
print('done with acupuncture diagnosis summary over last 24 months')

# for all encounters for acunpuncture users since FY21
icd_all = pd.read_sql("SELECT * FROM [OPCCCT_CIH].[dflt].[acupuncture_diagnoses_summary_all]", engine)
icd_all = get_ICD10_counts(icd_all, 1, 'icd10_count_all_all', 'icd10_patient_count_all_all')
print('done with diagnosis summary')

# for all encounters for acupuncture users in the last 24 months
# NOTE: This one must be summarized in SQL because it overloads the python memory
icd_all_24 = pd.read_sql("SELECT * FROM [OPCCCT_CIH].[dflt].[acupuncture_user_diagnoses_all_2_years]", engine)
print('done with diagnosis summary over last 24 months')

# merge all the datasets together
icd = icd_acu.merge(icd_acu_24, how='outer', on=['ICD10Code'])\
             .merge(icd_all, how='outer', on=['ICD10Code'])\
             .merge(icd_all_24, how='outer', on=['ICD10Code'])
icd
```

Get the % of total that each column represents

```python
icd_sum = icd.fillna(0)
cols = icd_sum.columns.drop('ICD10Code')
for col in cols:
    icd_sum[col+'_%'] = icd_sum[col]/sum(icd_sum[col])
icd_sum.to_csv('acupuncture_ICD10_counts.csv',index=False)
```

## Deciding to used ICD10 codes associated with acupuncture visits since FY21
There is a substantial difference in the distribution of ICD10 codes associated with acupuncture visits and those associated with patients who have ever received acupuncture

See what just the primary ICD10 codes are and whether there's a difference between traditional and BFA

```python
%%time
# For acupuncture specific encounters since FY21
icd = pd.read_sql("SELECT * FROM [OPCCCT_CIH].[dflt].[acupuncture_outpatient_diagnoses_acu_only] WHERE PrimarySecondary='P'", engine)
icd['trad'] = np.where(icd['Acupuncture_BFA']==1,0,1)
icd['trad_icd_count'] = icd.groupby(['ICD10Code'])['trad'].transform('sum')
icd['BFA_icd_count'] = icd.groupby(['ICD10Code'])['Acupuncture_BFA'].transform('sum')
icd = icd.fillna(0)
icd['trad_%'] = icd['trad_icd_count']/sum(icd['trad_icd_count'])
icd['BFA_%'] = icd['BFA_icd_count']/sum(icd['BFA_icd_count'])
icd = icd.filter(['ICD10Code','trad_icd_count','BFA_icd_count','trad_%','BFA_%']).drop_duplicates()
icd.to_csv('icd10_acupuncture_count_primary_code.csv',index=False)
```

### Get duration between visit info

```python
%%time
dur = visits.loc[(visits['Acupuncture_Traditional']==1) & (visits['Acupuncture_BFA']==0)]
dur['VisitDateTime'] = pd.to_datetime(dur['VisitDateTime'])

# Drop duplicate encounters sharing the same patient + timestamp
dur = dur.drop_duplicates(['PatientICN', 'VisitDateTime'])

# Aggregate to patient level
dur = dur.groupby('PatientICN').agg(first_visit=('VisitDateTime', 'min'),
                                        last_visit=('VisitDateTime', 'max'),
                                        visit_count=('VisitDateTime', 'count')).reset_index()

# Keep only patients with 2+ visits
dur = dur[dur['visit_count'] >= 2].copy()

dur['avg_days_between_visits'] = ((dur['last_visit'] - dur['first_visit']).dt.total_seconds()
                                    / 86400 / (dur['visit_count'] - 1))
```

```python
dur['avg_days_between_visits'].describe(percentiles=[.1,.25,.5,.75,.9,.99])
g = dur.loc[(dur['avg_days_between_visits']<=366) & (dur['avg_days_between_visits']>=1)]
print(g['avg_days_between_visits'].describe(percentiles=[.1,.25,.5,.75,.9,.99]))

plt.figure(figsize=(6, 6))
sns.boxplot(y='avg_days_between_visits', data=g, showfliers=0)
plt.title('Avg. Time between Acupuncture Visits (all types) per Patient')
#plt.xlabel('Fiscal Year')
plt.ylabel('Days between Encounters')
plt.xticks(rotation=45)
plt.tight_layout()
#plt.ylim(1,16)
plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda x, _: f'{x:,.0f}'))
#plt.gca().yaxis.set_major_locator(MultipleLocator(2))
plt.show()
```

```python
%%time
df = visits.copy()

df['VisitDateTime'] = pd.to_datetime(df['VisitDateTime'])

# Drop duplicate encounters sharing the same patient + timestamp
df = df.drop_duplicates(['PatientICN', 'VisitDateTime'])

# Sort so "1st visit" and ordering are correct
df = df.sort_values(['PatientICN', 'VisitDateTime'])

# Flag on the first encounter: was it a BFA?
first_is_bfa = (df.groupby('PatientICN').first()['Acupuncture_BFA'].eq(1).rename('first_was_bfa'))

# Did any visit have trad=1?
df['count_as_trad'] = np.where((df['Acupuncture_Traditional']==1) & (df['Acupuncture_BFA']==0),1,0)
any_trad = (df.groupby('PatientICN')['count_as_trad'].max().eq(1).rename('has_trad'))

# For "followed by", the trad must occur AFTER the first visit.
# First visit datetime per patient:
first_dt = df.groupby('PatientICN')['VisitDateTime'].transform('min')
trad_after_first = (df.loc[(df['Acupuncture_Traditional'] == 1) & (df['VisitDateTime'] > first_dt)]
    .groupby('PatientICN').size().gt(0)
    .reindex(df['PatientICN'].unique(), fill_value=False)
    .rename('trad_after_first'))

# Combine
result = pd.concat([first_is_bfa, trad_after_first], axis=1).reset_index()
result['trad_after_first'] = result['trad_after_first'].fillna(False)
result['bfa_then_trad'] = result['first_was_bfa'] & result['trad_after_first']

result = result.fillna(False)
print('Patients with BFA 1st',sum(result['first_was_bfa']))
print('Patients with subsequent Trad',sum(result['bfa_then_trad']))
print('%', sum(result['bfa_then_trad'])/sum(result['first_was_bfa']))
```

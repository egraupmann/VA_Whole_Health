## Basic library imports

```python
%%time
import pandas as pd
import numpy as np
import pyodbc
import urllib
from sqlalchemy import create_engine

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
```

## Establish the database connections with SQL Server
The recommended workflow is to make your data tables in Microsoft SQL Server and then simply access the joined/filtered outputs here. Trying to query the full datasets in Python is likely to create memory timeouts, depending on how much data you're trying to pull back. Creating and saving the desired manipulate in SQL Server also enables others to more easily access your data products.

```python
# establish the connection parameters
odbc_str = ("DRIVER={ODBC Driver 18 for SQL Server};"
            "SERVER=VHACDWA01.vha.med.va.gov,1433;"
            "DATABASE=OPCCCT_CIH;"
            "Trusted_Connection=yes;"
            "Encrypt=yes;"
            "TrustServerCertificate=no;")

# connect
params = urllib.parse.quote_plus(odbc_str)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

# read in the data
df = pd.read_sql("SELECT * FROM [OPCCCT_CIH].[dflt].[acupuncture_visits]", engine)
```
